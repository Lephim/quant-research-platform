```python
tickers = [
    "AAPL",
    "MSFT",
    "NVDA",
    "SPY",
    "BTC-USD"
]
```

```python
prices = yf.download(
    tickers,
    start="2018-01-01",
    end="2025-01-01"
)
```

```python
prices.head()
```

```python
close_prices = prices["Close"]

close_prices.head()
```

```python
normalized = (
    close_prices /
    close_prices.iloc[0]
) * 100
```

```python
plt.figure(figsize=(12,6))

plt.plot(normalized)

plt.title("Growth of $100 Investment")

plt.ylabel("Value")

plt.show()
```
